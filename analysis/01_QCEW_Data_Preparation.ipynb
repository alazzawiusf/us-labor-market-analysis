```python
import pandas as pd

url = "https://data.bls.gov/cew/data/files/2024/csv/2024_annual_singlefile.zip"

qcew = pd.read_csv(
    url,
    compression="zip",
    low_memory=False
)

print(qcew.shape)
print(qcew.columns.tolist())
qcew.head()
```

```python
sector_2024 = qcew[
    (qcew['agglvl_code'] == 54) &
    (qcew['own_code'].isin([1, 2, 3, 5]))
].copy()

print("Shape:", sector_2024.shape)

print(
    sector_2024[
        [
            'area_fips',
            'own_code',
            'industry_code',
            'annual_avg_emplvl',
            'total_annual_wages',
            'annual_avg_wkly_wage'
        ]
    ].head(30).to_string(index=False)
)
```

```python
# Aggregate ownership sectors into State × Industry
state_industry_2024 = (
    sector_2024
    .groupby(
        ['area_fips', 'year', 'industry_code'],
        as_index=False
    )
    .agg(
        establishments=('annual_avg_estabs', 'sum'),
        employment=('annual_avg_emplvl', 'sum'),
        total_wages=('total_annual_wages', 'sum')
    )
)

# Recalculate annual pay from totals
state_industry_2024['avg_annual_pay'] = (
    state_industry_2024['total_wages']
    / state_industry_2024['employment']
).round()

# Calculate weekly wage
state_industry_2024['avg_weekly_wage'] = (
    state_industry_2024['avg_annual_pay'] / 52
).round()

print("Shape:", state_industry_2024.shape)

# Alabama only
print(
    state_industry_2024[
        state_industry_2024['area_fips'] == '01000'
    ].to_string(index=False)
)
```

```python
# Sum employment across all Alabama industry sectors
alabama_sector_total = state_industry_2024.loc[
    state_industry_2024['area_fips'] == '01000',
    'employment'
].sum()

# Official QCEW Total Covered Employment for Alabama
alabama_official_total = qcew.loc[
    (qcew['area_fips'] == '01000') &
    (qcew['own_code'] == 0) &
    (qcew['industry_code'] == '10'),
    'annual_avg_emplvl'
].iloc[0]

print("Sum of industry sectors :", alabama_sector_total)
print("Official total covered  :", alabama_official_total)
print("Difference              :", alabama_sector_total - alabama_official_total)
print(
    "Difference %            :",
    round(
        (alabama_sector_total - alabama_official_total)
        / alabama_official_total * 100,
        4
    )
)
```

```python
# Total employment obtained from our industry-sector dataset
sector_totals = (
    state_industry_2024
    .groupby('area_fips', as_index=False)
    .agg(sector_employment=('employment', 'sum'))
)

# Official QCEW Total Covered Employment
official_totals = qcew[
    (qcew['agglvl_code'] == 50) &
    (qcew['own_code'] == 0) &
    (qcew['industry_code'] == '10')
][
    ['area_fips', 'annual_avg_emplvl']
].copy()

official_totals = official_totals.rename(
    columns={'annual_avg_emplvl': 'official_employment'}
)

# Compare
validation = sector_totals.merge(
    official_totals,
    on='area_fips',
    how='inner'
)

validation['difference'] = (
    validation['sector_employment']
    - validation['official_employment']
)

validation['difference_pct'] = (
    validation['difference']
    / validation['official_employment']
    * 100
)

print(validation.to_string(index=False))

print("\nMaximum absolute difference %:")
print(validation['difference_pct'].abs().max())

print("\nAverage absolute difference %:")
print(validation['difference_pct'].abs().mean())
```

```python
import pandas as pd
import numpy as np

years = range(2015, 2026)

all_state_totals = []
all_state_industry = []

for year in years:
    
    print(f"Processing {year}...")
    
    url = (
        f"https://data.bls.gov/cew/data/files/"
        f"{year}/csv/{year}_annual_singlefile.zip"
    )
    
    df = pd.read_csv(
        url,
        compression="zip",
        low_memory=False,
        dtype={
            "area_fips": str,
            "industry_code": str
        }
    )

    # -----------------------------------
    # 1. OFFICIAL STATE TOTALS
    # -----------------------------------
    
    state_totals = df[
        (df["agglvl_code"] == 50) &
        (df["own_code"] == 0) &
        (df["industry_code"] == "10")
    ].copy()

    state_totals = state_totals[
        [
            "area_fips",
            "year",
            "annual_avg_estabs",
            "annual_avg_emplvl",
            "total_annual_wages",
            "annual_avg_wkly_wage",
            "avg_annual_pay",
            "oty_annual_avg_emplvl_chg",
            "oty_annual_avg_emplvl_pct_chg",
            "oty_total_annual_wages_chg",
            "oty_total_annual_wages_pct_chg",
            "oty_avg_annual_pay_chg",
            "oty_avg_annual_pay_pct_chg"
        ]
    ]

    state_totals = state_totals.rename(
        columns={
            "annual_avg_estabs": "establishments",
            "annual_avg_emplvl": "employment",
            "total_annual_wages": "total_wages",
            "annual_avg_wkly_wage": "avg_weekly_wage"
        }
    )

    all_state_totals.append(state_totals)

    # -----------------------------------
    # 2. STATE × NAICS SECTOR
    # -----------------------------------

    sector = df[
        (df["agglvl_code"] == 54) &
        (df["own_code"].isin([1, 2, 3, 5]))
    ].copy()

    sector["is_suppressed"] = (
        sector["disclosure_code"] == "N"
    )

    state_industry = (
        sector
        .groupby(
            ["area_fips", "year", "industry_code"],
            as_index=False
        )
        .agg(
            establishments=("annual_avg_estabs", "sum"),
            employment=("annual_avg_emplvl", "sum"),
            total_wages=("total_annual_wages", "sum"),
            has_suppressed_data=("is_suppressed", "max")
        )
    )

    # Recalculate wage metrics
    state_industry["avg_annual_pay"] = np.where(
        state_industry["employment"] > 0,
        state_industry["total_wages"]
        / state_industry["employment"],
        np.nan
    )

    state_industry["avg_weekly_wage"] = (
        state_industry["avg_annual_pay"] / 52
    )

    state_industry["avg_annual_pay"] = (
        state_industry["avg_annual_pay"].round()
    )

    state_industry["avg_weekly_wage"] = (
        state_industry["avg_weekly_wage"].round()
    )

    all_state_industry.append(state_industry)

    print(
        f"  State totals: {len(state_totals):,} | "
        f"Industry rows: {len(state_industry):,}"
    )


# =======================================
# COMBINE ALL YEARS
# =======================================

state_totals = pd.concat(
    all_state_totals,
    ignore_index=True
)

state_industry = pd.concat(
    all_state_industry,
    ignore_index=True
)


print("\nDONE")
print("State totals shape:", state_totals.shape)
print("State × Industry shape:", state_industry.shape)

print("\nYears:")
print(state_totals["year"].value_counts().sort_index())
```

```python
# ==========================================
# QCEW QUALITY CONTROL
# ==========================================

print("=== 1. STATE CODES ===")

state_codes = sorted(
    state_totals["area_fips"].unique()
)

print(state_codes)
print("Number of areas:", len(state_codes))


print("\n=== 2. DUPLICATES ===")

total_duplicates = state_totals.duplicated(
    subset=["area_fips", "year"]
).sum()

industry_duplicates = state_industry.duplicated(
    subset=["area_fips", "year", "industry_code"]
).sum()

print("State total duplicates:", total_duplicates)
print("State-industry duplicates:", industry_duplicates)


print("\n=== 3. MISSING VALUES ===")

print("\nState totals:")
print(state_totals.isna().sum())

print("\nState × Industry:")
print(state_industry.isna().sum())


print("\n=== 4. INDUSTRY CODES BY YEAR ===")

for year in range(2015, 2026):

    codes = sorted(
        state_industry.loc[
            state_industry["year"] == year,
            "industry_code"
        ].unique()
    )

    print(year, codes)


print("\n=== 5. SUPPRESSION ===")

suppression = (
    state_industry
    .groupby("year")["has_suppressed_data"]
    .agg(["sum", "count"])
)

suppression["pct"] = (
    suppression["sum"]
    / suppression["count"]
    * 100
)

print(suppression)


print("\n=== 6. EMPLOYMENT SANITY CHECK ===")

print(
    state_totals[
        ["employment", "avg_annual_pay", "avg_weekly_wage"]
    ].describe()
)
```

```python
# ==========================================
# FINAL QCEW CLEANING
# ==========================================

# Remove Puerto Rico and U.S. Virgin Islands
exclude_territories = ["72000", "78000"]

state_totals = state_totals[
    ~state_totals["area_fips"].isin(exclude_territories)
].copy()

state_industry = state_industry[
    ~state_industry["area_fips"].isin(exclude_territories)
].copy()


# ------------------------------------------
# STATE NAMES
# ------------------------------------------

state_names = {
    "01000": "Alabama",
    "02000": "Alaska",
    "04000": "Arizona",
    "05000": "Arkansas",
    "06000": "California",
    "08000": "Colorado",
    "09000": "Connecticut",
    "10000": "Delaware",
    "11000": "District of Columbia",
    "12000": "Florida",
    "13000": "Georgia",
    "15000": "Hawaii",
    "16000": "Idaho",
    "17000": "Illinois",
    "18000": "Indiana",
    "19000": "Iowa",
    "20000": "Kansas",
    "21000": "Kentucky",
    "22000": "Louisiana",
    "23000": "Maine",
    "24000": "Maryland",
    "25000": "Massachusetts",
    "26000": "Michigan",
    "27000": "Minnesota",
    "28000": "Mississippi",
    "29000": "Missouri",
    "30000": "Montana",
    "31000": "Nebraska",
    "32000": "Nevada",
    "33000": "New Hampshire",
    "34000": "New Jersey",
    "35000": "New Mexico",
    "36000": "New York",
    "37000": "North Carolina",
    "38000": "North Dakota",
    "39000": "Ohio",
    "40000": "Oklahoma",
    "41000": "Oregon",
    "42000": "Pennsylvania",
    "44000": "Rhode Island",
    "45000": "South Carolina",
    "46000": "South Dakota",
    "47000": "Tennessee",
    "48000": "Texas",
    "49000": "Utah",
    "50000": "Vermont",
    "51000": "Virginia",
    "53000": "Washington",
    "54000": "West Virginia",
    "55000": "Wisconsin",
    "56000": "Wyoming"
}


# ------------------------------------------
# NAICS SECTOR NAMES
# ------------------------------------------

industry_names = {
    "11": "Agriculture, Forestry, Fishing and Hunting",
    "21": "Mining, Quarrying, and Oil and Gas Extraction",
    "22": "Utilities",
    "23": "Construction",
    "31-33": "Manufacturing",
    "42": "Wholesale Trade",
    "44-45": "Retail Trade",
    "48-49": "Transportation and Warehousing",
    "51": "Information",
    "52": "Finance and Insurance",
    "53": "Real Estate and Rental and Leasing",
    "54": "Professional, Scientific, and Technical Services",
    "55": "Management of Companies and Enterprises",
    "56": "Administrative and Support and Waste Management and Remediation Services",
    "61": "Educational Services",
    "62": "Health Care and Social Assistance",
    "71": "Arts, Entertainment, and Recreation",
    "72": "Accommodation and Food Services",
    "81": "Other Services (except Public Administration)",
    "92": "Public Administration",
    "99": "Unclassified"
}


# Add state names
state_totals["state"] = (
    state_totals["area_fips"].map(state_names)
)

state_industry["state"] = (
    state_industry["area_fips"].map(state_names)
)

# Add industry names
state_industry["industry"] = (
    state_industry["industry_code"].map(industry_names)
)


# ------------------------------------------
# ORDER COLUMNS
# ------------------------------------------

state_totals = state_totals[
    [
        "year",
        "area_fips",
        "state",
        "employment",
        "establishments",
        "total_wages",
        "avg_annual_pay",
        "avg_weekly_wage",
        "oty_annual_avg_emplvl_chg",
        "oty_annual_avg_emplvl_pct_chg",
        "oty_total_annual_wages_chg",
        "oty_total_annual_wages_pct_chg",
        "oty_avg_annual_pay_chg",
        "oty_avg_annual_pay_pct_chg"
    ]
]

state_industry = state_industry[
    [
        "year",
        "area_fips",
        "state",
        "industry_code",
        "industry",
        "employment",
        "establishments",
        "total_wages",
        "avg_annual_pay",
        "avg_weekly_wage",
        "has_suppressed_data"
    ]
]


# ------------------------------------------
# FINAL CHECK
# ------------------------------------------

print("State totals:", state_totals.shape)
print("State × Industry:", state_industry.shape)

print("\nStates per year:")
print(
    state_totals.groupby("year")["state"].nunique()
)

print("\nMissing state names:")
print(state_totals["state"].isna().sum())

print("\nMissing industry names:")
print(state_industry["industry"].isna().sum())
```
